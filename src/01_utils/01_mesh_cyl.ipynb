```python
# Required libraries

# Input shapefile
ruta_shp = DATA_MESH_FILE

# Load file and filter Castilla y Leon (CCA = 07)
gdf = gpd.read_file(ruta_shp).to_crs(epsg=25830)
gdf_cyl = gdf[gdf["CCA"] == "07"].copy()

print(gdf_cyl.info())

# Select only the required fields
cols_to_keep = ["CUSEC", "NMUN", "NPRO", "geometry"]
gdf_cyl = gdf_cyl[cols_to_keep]

# Create numeric ID optimized for Tableau relationships
gdf_cyl["Seccion_id"] = range(1, len(gdf_cyl) + 1)

# Output shapefile
salida = os.path.join(DATA_OUTPUTS_DIR, "Shapefiles", "cyl_2022.shp")
os.makedirs(os.path.dirname(salida), exist_ok=True)
gdf_cyl.to_file(salida)

print("✅ Optimized Castilla y León shapefile saved at:", salida)
```
